```python
import sys; sys.path.append("..")
import os
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score, f1_score

from src.data import load_adult
from src.models import build_models
from src.explain import load_tuned_params

os.makedirs("../results", exist_ok=True)
X_train, X_test, y_train, y_test = load_adult()
params = load_tuned_params()
```

```python
# Build models and save test predictions
models = build_models(X_train)
models["XGBoost (tuned)"] = build_models(X_train, xgb_params=params)["XGBoost (boosting)"]
order = ["Logistic Regression (baseline)", "Random Forest (bagging)",
         "XGBoost (boosting)", "XGBoost (tuned)", "Stacking (RF + XGB -> LR)"]

proba = {}
for name in order:
    proba[name] = models[name].fit(X_train, y_train).predict_proba(X_test)[:, 1]
    print("done:", name)
pd.DataFrame(proba).to_csv("../results/test_predictions.csv", index=False)
```
